```python
import numpy as np
```

### K-NN算法
1. 计算该点和所有点的距离
2. 按照距离进行排序
3. 获取距离最小的k个点
4. 确定前k个点类别的频率
5. 返回前k个点频率最高的类别作为预测

```python
def create_dataset():
    #这里的二维数据分别表示为 打斗次数 / 接吻次数
    #需要判断最终是爱情 / 动作 类型
    group = np.array([
        [1,100],
        [5,89],
        [108,5],
        [115,8]
    ])
    labels = ['爱情','爱情','动作','动作']
    return group,labels
```

`距离公式计算`
$$
  \sqrt{(x_1-x_0)^2 + (y_1-y_0)^2}
$$

```python
#paraments: 输入, 数据集, 标签, k
def classify(x, dataset, labels, k):
    dataset_size = dataset.shape[0]
    #让输入数组变成datasetSize个
    diff_mat = np.tile(x, (dataset_size,1)) - dataset
    sq_diff_mat = diff_mat**2
    #按照列来求和 降为一维
    sq_distance = sq_diff_mat.sum(axis=1)
    distances = sq_distance**0.5
    #对index进行排序, 返回排序后的索引
    #如[5 , 1 , 3 , 8] => [1 , 2 , 0 , 3]
    sorted_distances_index = distances.argsort()
    class_count = {}
    for i in range(k):
        #获取label,核心还是上面对索引进行排序
        vote_label = labels[sorted_distances_index[i]]
        class_count[vote_label] = class_count.get(vote_label,0) + 1
    sorted_class_count = sorted(class_count.items() , key = lambda item:item[1] , reverse=True)
    #返回第一个的类别
    return sorted_class_count[0][0]
```

```python
dataset , labels = create_dataset()
test = [101,20]
test_class = classify(test,dataset,labels,3)
print(test_class)
```
